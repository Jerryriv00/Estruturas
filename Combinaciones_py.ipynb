(n=10)

(r=8)

C(10, 8) = 10! / (8! × (10 − 8)!)

C(10, 8) = 10! / (8! × 2!)

C(10, 8) = 45

##1.6. ¿De cuántas formas es posible distribuir 10 monedas (idénticas) entre cinco niños si:

a) no hay restricciones?  

Fórmula:

C(n + k − 1, k − 1)

Datos:

Monedas = 10

Niños = 5

Cálculo:

C(10 + 5 − 1, 5 − 1) = C(14, 4)

C(14, 4) = 14! / (4! × 10!)

C(14, 4) = 1001

b) cada niño recibe al menos una moneda?  

Primero, entregamos una moneda a cada niño.

Monedas restantes: 10 − 5 = 5

Ahora distribuimos las 5 monedas restantes entre los 5 niños sin restricciones.

Cálculo:

C(5 + 5 − 1, 5 − 1) = C(9, 4)

C(9, 4) = 9! / (4! × 5!)

C(9, 4) = 126

c) el niño mayor recibe al menos dos monedas?

Primero, entregamos 2 monedas al niño mayor.

Monedas restantes:

10 − 2 = 8

Ahora distribuimos las 8 monedas restantes entre los 5 niños sin restricciones.

Cálculo:

C(8 + 5 − 1, 5 − 1) = C(12, 4)

C(12, 4) = 12! / (4! × 8!)

C(12, 4) = 495

```python
#3
from math import factorial, comb
from math import factorial
from collections import Counter

texto=input("Ingrese texto: ").upper()
r=1
for c in Counter(texto).values():
    r *=factorial(c)
print("Las disposiciones son :", factorial(len(texto))//r)
```

El programa permite ingresar una palabra y calcular cuántas formas diferentes existen de ordenar sus letras teniendo en cuenta las repeticiones. La variable texto guarda la palabra ingresada, r acumula los factoriales de las letras repetidas y c representa la cantidad de veces que aparece cada letra. Para realizar el cálculo se utilizan Counter() para contar las repeticiones, factorial() para calcular los factoriales y len() para obtener la cantidad de letras.

```python
#4
from math import factorial, comb
from collections import Counter
texto="TALLAHASSEE"
resto= texto.replace("A", "")
r=1
for c in Counter(resto).values():
    r *=factorial(c)
print("Las disposiciones son:", factorial(len(resto))//r * comb(len(resto)+1,texto.count("A")))
```

Este programa calcula cuántas formas diferentes existen de ordenar las letras de TALLAHASSEE sin que las letras A queden juntas. La variable texto guarda la palabra original, resto almacena la palabra sin las letras A, r acumula los factoriales de las letras repetidas y c representa las repeticiones de cada letra restante. Se utilizan replace() para quitar las A, Counter() para contar las letras repetidas, factorial() para realizar los cálculos y comb() para elegir los espacios donde se colocarán las A.

```python
#5
from math import factorial, comb
from collections import Counter

texto=input("Ingrese texto: ").upper()
letra=input("Ingrese letra: ").upper()
resto=texto.replace(letra, "")
k=texto.count(letra)
r=1
for c in Counter(resto).values():
    r *=factorial(c)
print("las disposiciones son:", 0 if k >len(resto)+1 else factorial(len(resto))//r*comb(len(resto)+1,k))
```

El programa permite ingresar una palabra y elegir una letra para calcular cuántas formas existen de ordenar sus letras sin que la letra elegida aparezca dos veces seguidas. La variable texto guarda la palabra, letra almacena la letra seleccionada, resto contiene la palabra sin esa letra, k indica cuántas veces aparece la letra elegida, r acumula los factoriales de las letras repetidas y c representa las repeticiones de las letras restantes. Se utilizan replace() para separar la letra seleccionada, count() para contar sus apariciones, Counter() para contar las demás letras y factorial() y comb() para calcular las disposiciones posibles.
